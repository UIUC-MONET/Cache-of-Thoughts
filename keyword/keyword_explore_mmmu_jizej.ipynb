```python
test_keyword_dir = Path('/Users/jizej/Workspaces/UOUO/keyword/mmmu/test_keyword/')
val_keyword_dir = Path('/Users/jizej/Workspaces/UOUO/keyword/mmmu/val_keyword/')
```

```python
test_keywords = []
# load keywords from each txt file
count = 0
for kw_file in test_keyword_dir.glob('*.txt'):
    with open(kw_file, 'r') as f:
        # for each line
        for line in f:
            if line.strip() == '':
                continue
            count += 1
            # load comma separated keywords, strip each keyword
            keywords = [kw.strip() for kw in line.strip().split(',')]
            test_keywords.append(keywords)
```

```python
from collections import Counter
# count freq of each keyword
keyword_freq = Counter()
for keywords in test_keywords:
    for kw in keywords:
        keyword_freq[kw] += 1
```

```python
for each in keyword_freq.most_common(2000):
    print(each)
```
